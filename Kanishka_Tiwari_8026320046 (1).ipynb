LAB 4.
Develop the salary and Education data in .csv file, read the file, make the vectors, compute the Eigen values, Covariance matrix, and apply PCA to reduce the dimensions of data.

```python
# LAB 4: Salary and Education using PCA
# Basic Python Only (Without NumPy, Pandas or Scikit-learn)

# Step 1: Create CSV file
f = open("Exp4.csv", "w")

f.write("Education,Salary\n")
f.write("10,20000\n")
f.write("12,25000\n")
f.write("14,30000\n")
f.write("16,40000\n")
f.write("18,50000\n")
f.write("20,60000\n")

f.close()

print("CSV file created successfully!")

# Step 2: Read CSV file
f = open("Exp4.csv", "r")
data = f.readlines()
f.close()

education = []
salary = []

for row in data[1:]:
    values = row.strip().split(",")
    education.append(float(values[0]))
    salary.append(float(values[1]))

n = len(education)

print("\nEducation Vector:", education)
print("Salary Vector:", salary)

# Step 3: Calculate mean
mean_e = sum(education) / n
mean_s = sum(salary) / n

print("\nMean Education:", mean_e)
print("Mean Salary:", mean_s)

# Step 4: Center the data
E = []
S = []

for i in range(n):
    E.append(education[i] - mean_e)
    S.append(salary[i] - mean_s)

print("\nCentered Education:", E)
print("Centered Salary:", S)

# Step 5: Calculate covariance matrix
var_e = sum(x * x for x in E) / (n - 1)
var_s = sum(x * x for x in S) / (n - 1)
cov_es = sum(E[i] * S[i] for i in range(n)) / (n - 1)

print("\nCovariance Matrix:")
print([var_e, cov_es])
print([cov_es, var_s])

# Step 6: Standardize the data
std_e = var_e ** 0.5
std_s = var_s ** 0.5

Z1 = [x / std_e for x in E]
Z2 = [x / std_s for x in S]

print("\nStandardized Education:", Z1)
print("Standardized Salary:", Z2)

# Step 7: Calculate standardized covariance matrix
a = sum(x * x for x in Z1) / (n - 1)
b = sum(Z1[i] * Z2[i] for i in range(n)) / (n - 1)
d = sum(x * x for x in Z2) / (n - 1)

print("\nStandardized Covariance Matrix:")
print([a, b])
print([b, d])

# Step 8: Calculate eigenvalues
trace = a + d
det = a * d - b * b

discriminant = trace * trace - 4 * det

if discriminant < 0 and discriminant > -0.000000001:
    discriminant = 0

root = discriminant ** 0.5

lambda1 = (trace + root) / 2
lambda2 = (trace - root) / 2

print("\nEigenvalue 1:", lambda1)
print("Eigenvalue 2:", lambda2)

# Step 9: Calculate first eigenvector
x = b
y = lambda1 - a

length = (x * x + y * y) ** 0.5

if length > 0.000000001:
    x = x / length
    y = y / length
else:
    x = 1
    y = 0

print("\nFirst Eigenvector:", [x, y])

# Step 10: Apply PCA
reduced_data = []

for i in range(n):
    score = Z1[i] * x + Z2[i] * y
    reduced_data.append(score)

print("\nPCA Reduced Data (2D to 1D):")

for i in range(n):
    print("Person", i + 1, ":", reduced_data[i])

# Step 11: Calculate variance retained
total = lambda1 + lambda2

if total != 0:
    retained = (lambda1 / total) * 100
else:
    retained = 0

print("\nVariance Retained:", retained, "%")

print("\nPCA Dimensionality Reduction Completed Successfully!")
```

LAB 8.  Real-world example: Student heights, measurement errors, manufacturing dimensions, etc. (i) Apply CLT to above real dataset, verify SE = 𝜎 √𝑛 , experimentally. (ii) Generate and plot a normal distribution.

```python

import random
import math
import matplotlib.pyplot as plt

# Student Height Dataset
heights = [
    165, 170, 168, 175, 160,
    172, 169, 180, 166, 174,
    163, 171, 177, 168, 164,
    173, 176, 169, 162, 178,
    170, 167, 175, 181, 166,
    172, 164, 179, 168, 173
]

# Calculate Mean
mu = sum(heights) / len(heights)

# Calculate Standard Deviation
variance = sum((x - mu) ** 2 for x in heights) / len(heights)
sigma = math.sqrt(variance)

print("Mean =", mu)
print("Standard Deviation =", sigma)

# PART 1: CENTRAL LIMIT THEOREM

sample_sizes = [5, 10, 20, 30]

for n in sample_sizes:
    means = []

    for i in range(1000):
        sample = random.choices(heights, k=n)
        sample_mean = sum(sample) / n
        means.append(sample_mean)

    # Calculate Theoretical Standard Error
    theoretical_se = sigma / math.sqrt(n)

    # Calculate Experimental Standard Error
    mean_of_means = sum(means) / len(means)

    exp_variance = sum(
        (x - mean_of_means) ** 2 for x in means
    ) / (len(means) - 1)

    experimental_se = math.sqrt(exp_variance)

    print("\nSample Size:", n)
    print("Theoretical SE:", theoretical_se)
    print("Experimental SE:", experimental_se)

    # Plot Sampling Distribution
    plt.hist(means, bins=20, edgecolor="black")
    plt.title("CLT for Sample Size " + str(n))
    plt.xlabel("Sample Mean")
    plt.ylabel("Frequency")
    plt.show()

# PART 2: NORMAL DISTRIBUTION

x_values = []
y_values = []

x = mu - 4 * sigma

while x <= mu + 4 * sigma:
    y = (1 / (sigma * math.sqrt(2 * math.pi))) * math.exp(
        -((x - mu) ** 2) / (2 * sigma ** 2)
    )

    x_values.append(x)
    y_values.append(y)

    x = x + 0.5

# Plot Normal Distribution
plt.hist(heights, bins=10, density=True, edgecolor="black")
plt.plot(x_values, y_values)

plt.title("Normal Distribution of Student Heights")
plt.xlabel("Height")
plt.ylabel("Probability Density")
plt.show()
```

(b) Think about two teaching methods and compare the average EST marks using t-test.

```python

# LAB 9 (b)
# Compare Two Teaching Methods Using t-test

# Marks of students using Method A and Method B
A = [65, 70, 68, 72, 60]
B = [75, 80, 78, 85, 74]

n1 = len(A)
n2 = len(B)

# Step 1: Calculate Mean of Method A
sum1 = 0

for i in range(n1):
    sum1 = sum1 + A[i]

mean1 = sum1 / n1

# Step 2: Calculate Mean of Method B
sum2 = 0

for i in range(n2):
    sum2 = sum2 + B[i]

mean2 = sum2 / n2

print("Mean of Method A:", mean1)
print("Mean of Method B:", mean2)

# Step 3: Calculate Squared Deviations
sq1 = 0
sq2 = 0

for i in range(n1):
    sq1 = sq1 + (A[i] - mean1) ** 2

for i in range(n2):
    sq2 = sq2 + (B[i] - mean2) ** 2

# Step 4: Calculate Sample Variances
var1 = sq1 / (n1 - 1)
var2 = sq2 / (n2 - 1)

print("Variance A:", var1)
print("Variance B:", var2)

# Step 5: Calculate Pooled Variance
sp = ((n1 - 1) * var1 + (n2 - 1) * var2) / (n1 + n2 - 2)

print("Pooled Variance:", sp)

# Step 6: Calculate t-value
t = (mean1 - mean2) / ((sp * (1/n1 + 1/n2)) ** 0.5)

print("T-value:", t)

# Step 7: Compare with Critical t-value
critical_t = 2.306

if abs(t) > critical_t:
    print("Reject H0: Significant difference exists")
else:
    print("Accept H0: No significant difference")
```

LAB 10.
Discussion of Type I and type 2 error on binary classification problem. Real Situation Test Says: Healthy Test says: Disease Actual reading: Healthy correct Type 1 error Actual reading: Diseased Type II error correct  Problem: Medical screening test Suppose a disease- screening test is applied to 1000 people then following observations are made
(a) 100 actually have the disease
(b) 900 are healthy but test says diseased.
(c) Test incorrectly identifies 5% of healthy people are diseased.
 (d) Test fails to detect 10% of diseased people.  Test exercise, simulate Airport Security and following Hypothesis should be made from the empirical investigation. H₀: Passenger does NOT have a prohibited item. H₁: Passenger HAS a prohibited item. Develop the problem by your own and simulate it for 10000 passengers. (i) Number of Type 1 and 2 errors. (ii) Connect the aforementioned problem with confusion matrix.

```python

import random

random.seed(42)

total = 10000

# Actual passenger status
actual = []
predicted = []

# Generate 10000 passengers
for i in range(total):

    if random.random() < 0.02:
        actual.append(1)  # Prohibited item
    else:
        actual.append(0)  # Safe

# Security screening
for i in range(total):

    if actual[i] == 0:

        if random.random() < 0.05:
            predicted.append(1)  # Type 1 error
        else:
            predicted.append(0)  # Correct

    else:

        if random.random() < 0.10:
            predicted.append(0)  # Type 2 error
        else:
            predicted.append(1)  # Correct

# Confusion matrix
TN = 0
FP = 0
FN = 0
TP = 0

for i in range(total):

    if actual[i] == 0 and predicted[i] == 0:
        TN = TN + 1

    elif actual[i] == 0 and predicted[i] == 1:
        FP = FP + 1

    elif actual[i] == 1 and predicted[i] == 0:
        FN = FN + 1

    else:
        TP = TP + 1

# Display results
print("AIRPORT SECURITY SIMULATION")
print("Total Passengers:", total)

print("\nConfusion Matrix")
print("                 Predicted")
print("              Safe  Prohibited")
print("Actual Safe  ", TN, " ", FP)
print("Actual Item  ", FN, " ", TP)

print("\nType 1 Error:", FP)
print("Type 2 Error:", FN)

print("\nType 1 Error Rate:", FP / (TN + FP) * 100)
print("Type 2 Error Rate:", FN / (FN + TP) * 100)

accuracy = (TN + TP) / total * 100

print("Accuracy:", accuracy, "%")
```

LAB 11. Perform One tailed vs Two-tailed test: A sample of 25 energy efficient light bulbs has 
mean lifetime of 1050 hrs with sample standard deviation of 80 hours. The 
manufacturer claims the mean lifetime is 1000 hrs. 
(a) Perform a two-tailed test one sample t-test with 𝛼=0.05 using 
scipy.stats.ttest_1samp. 
(b) State the null and alternate hypothesis, the test statistics, p-values  on the basis 
of certain readings by using binary classification problems. 

```python

from scipy.stats import t

# Given values
sample_size = 25
sample_mean = 1050
sample_std = 80
population_mean = 1000
alpha = 0.05

# Calculate t-statistic
standard_error = sample_std / (sample_size ** 0.5)

t_statistic = (sample_mean - population_mean) / standard_error

# Degrees of freedom
df = sample_size - 1

# Calculate two-tailed p-value
p_value = 2 * t.sf(abs(t_statistic), df)

print("Sample Mean:", sample_mean)
print("Population Mean:", population_mean)
print("T-statistic:", round(t_statistic, 4))
print("P-value:", round(p_value, 4))

# Decision
if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python

# Linear Regression using Sales and Year

# Randomly created sales data
years = [2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]

sales = [100, 120, 140, None, 180, None, 220, 240]

print("Original Sales Data:")
for i in range(len(years)):
    print(years[i], ":", sales[i])

# Calculate average of available sales
total_sales = 0
count = 0

for value in sales:
    if value is not None:
        total_sales = total_sales + value
        count = count + 1

mean_sales = total_sales / count

# Calculate average year
mean_year = sum(years) / len(years)

# Calculate slope
numerator = 0
denominator = 0

for i in range(len(years)):
    if sales[i] is not None:
        numerator += (years[i] - mean_year) * (sales[i] - mean_sales)
        denominator += (years[i] - mean_year) ** 2

slope = numerator / denominator

# Calculate intercept
intercept = mean_sales - (slope * mean_year)

print("\nLinear Regression Equation:")
print("Sales =", round(slope, 2), "* Year +", round(intercept, 2))

# Fill missing sales values
for i in range(len(sales)):
    if sales[i] is None:
        predicted_sales = slope * years[i] + intercept
        sales[i] = round(predicted_sales, 2)

# Display final data
print("\nSales Data After Filling Missing Values:")

for i in range(len(years)):
    print(years[i], ":", sales[i])
```

LAB 13. WAP by using real Kaggle dataset for monthly energy consumption dataset and 
compute the following: 
(A) Compute and plot the 3 and 6-month simple moving average data. 
(B) Compute the weighted moving average and compare the smoothness against the 
simple moving average. 
(C) Comments on how the choice of window size effects the trade-off between the 
noise reduction and lag. 

```python

import csv

# Read the energy consumption dataset
monthly_data = {}


# Link to downloae the exact kaggle dataset, download this in the form of ZIP file & extract all- https://www.kaggle.com/datasets/uciml/electric-power-consumption-data-set?utm_source=chatgpt.com
with open("household_power_consumption/household_power_consumption.txt", "r") as file:    reader = csv.DictReader(file, delimiter=";")

    for row in reader:
        date = row["Date"]
        power = row["Global_active_power"]

        # Skip missing values
        if power == "?":
            continue

        # Extract year and month
        day, month, year = date.split("/")
        month_key = year + "-" + month

        # Convert power into kWh
        energy = float(power) / 60

        if month_key not in monthly_data:
            monthly_data[month_key] = 0

        monthly_data[month_key] += energy


# Sort monthly data by date
months = sorted(monthly_data.keys())

consumption = []

for month in months:
    consumption.append(monthly_data[month])


print("Total Months:", len(months))
print("\nMonthly Energy Consumption:")

for i in range(len(months)):
    print(months[i], ":", round(consumption[i], 2), "kWh")


# Calculate Simple Moving Average
def simple_moving_average(data, window):

    result = []

    for i in range(window - 1, len(data)):
        total = 0

        for j in range(i - window + 1, i + 1):
            total += data[j]

        average = total / window
        result.append(average)

    return result


# Calculate Weighted Moving Average
def weighted_moving_average(data, window):

    result = []

    for i in range(window - 1, len(data)):
        total = 0
        weight_total = 0

        for j in range(window):
            weight = j + 1
            total += data[i - window + 1 + j] * weight
            weight_total += weight

        average = total / weight_total
        result.append(average)

    return result


# Calculate 3 and 6 month averages
sma_3 = simple_moving_average(consumption, 3)
sma_6 = simple_moving_average(consumption, 6)

wma_3 = weighted_moving_average(consumption, 3)
wma_6 = weighted_moving_average(consumption, 6)


# Display moving average results
print("\n3 Month Simple Moving Average:", [round(x, 2) for x in sma_3])
print("\n6 Month Simple Moving Average:", [round(x, 2) for x in sma_6])

print("\n3 Month Weighted Moving Average:", [round(x, 2) for x in wma_3])
print("\n6 Month Weighted Moving Average:", [round(x, 2) for x in wma_6])


# Compare smoothness using standard deviation of changes
def calculate_smoothness(data):

    changes = []

    for i in range(1, len(data)):
        changes.append(data[i] - data[i - 1])

    if len(changes) == 0:
        return 0

    mean = sum(changes) / len(changes)

    variance = 0

    for value in changes:
        variance += (value - mean) ** 2

    variance = variance / len(changes)

    return variance ** 0.5


print("\nSmoothness Comparison")
print("SMA 3:", round(calculate_smoothness(sma_3), 2))
print("SMA 6:", round(calculate_smoothness(sma_6), 2))
print("WMA 3:", round(calculate_smoothness(wma_3), 2))
print("WMA 6:", round(calculate_smoothness(wma_6), 2))


# Simple text-based plot for the last 12 months
print("\nMonthly Consumption Plot (Last 12 Months)")

last_months = months[-12:]
last_values = consumption[-12:]

maximum = max(last_values)

for month, value in zip(last_months, last_values):
    bars = int((value / maximum) * 40)
    print(month, "|" + "*" * bars)


# Display moving average comparison
print("\nMoving Average Comparison (Last 12 Months)")

for i in range(max(0, len(months) - 12), len(months)):
    print("\nMonth:", months[i])
    print("Actual:", round(consumption[i], 2))

    if i >= 2:
        print("SMA 3:", round(sma_3[i - 2], 2))
        print("WMA 3:", round(wma_3[i - 2], 2))

    if i >= 5:
        print("SMA 6:", round(sma_6[i - 5], 2))
        print("WMA 6:", round(wma_6[i - 5], 2))
```

LAB 14. WAP to develop the autoregressive model for climate forecasting data. Plot ACF and 
PACF to identify a suitable AR of order p, forecast the climate reading for next 15 
days and compute the RMSE. 

```python

import csv
import math

# Reading climate data from CSV file
temperature = []

#USE THIS link to get the kaggle data set - https://www.kaggle.com/sumanthvrao/daily-climate-time-series-data?utm_source=chatgpt.com
with open("climate_data.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        try:
            value = float(row["temperature"])
            temperature.append(value)
        except:
            continue

print("Total climate readings:", len(temperature))


# Function to calculate mean
def find_mean(data):
    return sum(data) / len(data)


# Function to calculate autocorrelation (ACF)
def calculate_acf(data, lag):
    mean = find_mean(data)

    numerator = 0
    denominator = 0

    for i in range(len(data)):
        denominator += (data[i] - mean) ** 2

    for i in range(lag, len(data)):
        numerator += (data[i] - mean) * (data[i - lag] - mean)

    if denominator == 0:
        return 0

    return numerator / denominator


# Function to solve equations using Gaussian elimination
def solve_equations(matrix, values):
    n = len(values)

    for i in range(n):
        pivot = matrix[i][i]

        if abs(pivot) < 0.0000001:
            return [0] * n

        for j in range(i, n):
            matrix[i][j] /= pivot

        values[i] /= pivot

        for k in range(n):
            if k != i:
                factor = matrix[k][i]

                for j in range(i, n):
                    matrix[k][j] -= factor * matrix[i][j]

                values[k] -= factor * values[i]

    return values


# Function to calculate PACF using regression
def calculate_pacf(data, lag):
    if lag == 0:
        return 1

    size = len(data) - lag

    if size <= lag:
        return 0

    matrix = []
    values = []

    for i in range(lag):
        row = []

        for j in range(lag):
            row.append(calculate_acf(data, abs(i - j)))

        matrix.append(row)

    for i in range(1, lag + 1):
        values.append(calculate_acf(data, i))

    coefficients = solve_equations(matrix, values)

    return coefficients[-1]


# Plotting ACF and PACF using simple text bars
def show_plot(title, values):
    print("\n", title)
    print("-" * 45)

    for lag, value in enumerate(values, start=1):
        bar_length = int(abs(value) * 40)
        bar = "*" * bar_length

        print("Lag", lag, "|", bar, round(value, 3))


# Calculate ACF and PACF
max_lag = min(10, len(temperature) // 4)

acf_values = []
pacf_values = []

for lag in range(1, max_lag + 1):
    acf_values.append(calculate_acf(temperature, lag))
    pacf_values.append(calculate_pacf(temperature, lag))

show_plot("ACF PLOT", acf_values)
show_plot("PACF PLOT", pacf_values)


# Selecting AR order using PACF
confidence_limit = 1.96 / math.sqrt(len(temperature))

p = 1

for lag, value in enumerate(pacf_values, start=1):
    if abs(value) > confidence_limit:
        p = lag

print("\nSelected AR order (p):", p)


# Function to train AR model
def train_ar_model(data, p):
    matrix = []
    values = []

    for i in range(p, len(data)):
        row = [1]

        for j in range(1, p + 1):
            row.append(data[i - j])

        matrix.append(row)
        values.append(data[i])

    size = p + 1

    equation_matrix = []
    equation_values = []

    for i in range(size):
        row = []

        for j in range(size):
            total = 0

            for k in range(len(matrix)):
                total += matrix[k][i] * matrix[k][j]

            row.append(total)

        equation_matrix.append(row)

        total = 0

        for k in range(len(matrix)):
            total += matrix[k][i] * values[k]

        equation_values.append(total)

    coefficients = solve_equations(
        equation_matrix, equation_values
    )

    return coefficients


# Function to predict next value
def predict_next(history, coefficients, p):
    prediction = coefficients[0]

    for i in range(1, p + 1):
        prediction += coefficients[i] * history[-i]

    return prediction


# Splitting data into training and testing
train_size = len(temperature) - 15

train_data = temperature[:train_size]
test_data = temperature[train_size:]

# Train the model
coefficients = train_ar_model(train_data, p)

print("\nAR Model Coefficients:", coefficients)


# Forecasting test data
history = train_data.copy()
predictions = []

for actual in test_data:
    predicted = predict_next(history, coefficients, p)
    predictions.append(predicted)
    history.append(actual)


# Calculating RMSE
squared_error = 0

for i in range(len(test_data)):
    error = test_data[i] - predictions[i]
    squared_error += error ** 2

rmse = math.sqrt(squared_error / len(test_data))

print("\nRMSE:", round(rmse, 3))


# Forecasting next 15 days using complete dataset
final_coefficients = train_ar_model(temperature, p)

history = temperature.copy()
future_predictions = []

for day in range(1, 16):
    next_value = predict_next(history, final_coefficients, p)

    future_predictions.append(next_value)
    history.append(next_value)


# Displaying 15-day forecast
print("\nNEXT 15 DAYS CLIMATE FORECAST")
print("--------------------------------")

for day, value in enumerate(future_predictions, start=1):
    print("Day", day, ":", round(value, 2))


print("\nClimate forecasting completed successfully!")
```
